```python
# Create 1 new example that demonstrates you understand the problem. 
# Trace/walkthrough 1 example that your partner made and explain it
'''Example: Find the missing number in the given list 
# input = [2,3,1,2,0,6,4,8]
# output = [5,7]'''
# Example that my partner made 
'''Input: lst = [0, 1, 2]
Output: [-1]
My partner created a list of integers ranging from 0 to 2. The problem is to find the missing number/numbers
ranging from 0-n and return it as list, otherwise retun -1. In the given example, since no number is missing,
therefore, the output is [-1].'''
```

```python
#Copy the solution your partner wrote.
def missing_num(nums: list) -> int:
  # Convert to a set so later operators on nums have less time comlexity
    nums_set = set(nums) # O(n)\n"
    # Initialize a set to store the missing numbers
    missing_num_set = set() # O(1)\n
  # Find the highest number(n) in the given set and generate the range from 0 to n(inclusive);
  # if any number in the range is not present in the \n",
   # nums set, add it to the missing_num_set
    for i in range(max(nums_set)+1): # O(n+1) , O(n)
      if i not in nums_set:          # O(1)
         missing_num_set.add(i)       # O(1)
  
  # if the missing_num_set is not empty, then convert to a list and return it, else return -1
    if missing_num_set:  #O(1)
      return list(missing_num_set) #O(1)
    
    return -1 # O(1)
    
    print(missing_num([0,2]))
    print(missing_num([5,0,1]))
    print(missing_num([6, 8, 2, 3, 5, 7, 0, 1, 10]))
    print(missing_num([0,1,2]))
    print(missing_num([9,9,9]))
```
