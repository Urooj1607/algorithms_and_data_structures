```python


# Definition for a binary tree node.
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def bt_path(root):
    def dfs(node, path, result):
        if node is None:
            return
        
        path.append(node.val)

        # If it's a leaf node, append the path to result
        if node.left is None and node.right is None:
            result.append(list(path))  # Make a copy of the path
        
        # Recursively call for left and right subtree
        dfs(node.left, path, result)
        dfs(node.right, path, result)

        path.pop()  # Backtrack

    result = []
    dfs(root, [], result)
    return result

```

```python
class TreeNode(object):
  def __init__(self, val = 0, left = None, right = None):
   self.val = val
   self.left = left
   self.right = right

def br_path(root: TreeNode) -> list[list[int]]:
        if not root:
            return []
        if not root.left and not root.right:
            return [[root.val]]
        
        paths = []

        if root.left:
            left_paths = br_path(root.left)
            for path in left_paths:
                paths.append([root.val] + path)

        if root.right:
            right_paths = br_path(root.right)
            for path in right_paths:
                paths.append([root.val] + path)
        return paths


    
        
```

```python
def ft_path(root: TreeNode) -> list[list[int]]:
    if not root:
        return []
    
    result = [[root.val]]  # Start with root node as the first list
    
    left_paths = ft_path(root.left)   # Recursively get left subtree paths
    right_paths = ft_path(root.right) # Recursively get right subtree paths
    
    return result + left_paths + right_paths  # Combine root, left, and right lists

# Example Binary Tree
root = TreeNode(1)
root.left = TreeNode(2)
root.right = TreeNode(3)
root.left.left = TreeNode(4)
root.left.right = TreeNode(5)

print(ft_path(root))
```
